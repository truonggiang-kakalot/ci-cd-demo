```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
giobai = np.array([[1,1],[1,2],[2,1],[2,2],[3,4],[4,6],[5,8],[5,9]])
diem = np.array([2,3,3,4,7,8,9,10]).reshape(-1,1)
```

```python
lr=linear_model.LinearRegression()
lr.fit(giobai,diem)
print('he so a =', lr.coef_)
print('he so b =', lr.intercept_)
fig = plt.figure(figsize=(6,6))
dt=fig.add_subplot(111,projection='3d')
plt.scatter(giobai[:,0],giobai[:,1],diem,color ='red')
dt.set_xlabel('gio hoc')
dt.set_ylabel('so bai')
dt.set_zlabel('diem')
dt.set_title('Hoi quy 3d')
lr=linear_model.LinearRegression()
lr.fit(giobai,diem)
print('he so a =', lr.coef_)
print('he so b =', lr.intercept_)
gio=np.linspace(min(giobai[:,0]),max(giobai[:,0]))
bai=np.linspace(min(giobai[:,1]),max(giobai[:,1]))
x1,x2=np.meshgrid(gio,bai)
y=lr.coef_[0][0]*x1 + lr.coef_[0][1]*x2 + lr.intercept_[0]
dt.plot_surface(x1,x2,y,color='cyan', alpha=0.5, edgecolor='k',label ='Mặt phẳng hồi quy')
dt.legend()
xy_test = np.array([3,6])
kq_test=lr.coef_[0][0]*xy_test[0] + lr.coef_[0][1]*xy_test[1] + lr.intercept_[0]
print(kq_test)
```
